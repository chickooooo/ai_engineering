```python
import anthropic
```

```python
# Constants
MODEL = "claude-haiku-4-5-20251001"
```

```python
# Anthropic client
client = anthropic.Anthropic()
```

```python
# Make LLM call
message = client.messages.create(
    model=MODEL,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "Tell me a joke",
        }
    ],
)
```

```python
# Inspect response
# See `project_0/sample_response.json`
message.model_dump_json()
```

```python
# Print response
for block in message.content:
    if block.type == "text":
        print(block.text)
```
